```python
import pandas as pd
from pathlib import Path

# Original CSV ka existence check
assert RAW_CSV_PATH.exists(), "Original CSV file not found!"

# Original file ka initial size record
ORIGINAL_FILE_SIZE = RAW_CSV_PATH.stat().st_size

# Original CSV read karna
df_raw = pd.read_csv(RAW_CSV_PATH)

# Separate working copy
df = df_raw.copy(deep=True)

# Original file ka size dobara check
CURRENT_FILE_SIZE = RAW_CSV_PATH.stat().st_size

print("ORIGINAL DATASET LOADED SUCCESSFULLY!")

print("\nOriginal Rows:", len(df_raw))
print("Working Copy Rows:", len(df))

print("\nOriginal Columns:", len(df_raw.columns))
print("Working Copy Columns:", len(df.columns))

print("\nOriginal File Size Unchanged:",
      ORIGINAL_FILE_SIZE == CURRENT_FILE_SIZE)

print("\nOriginal Dataset Preview:")
display(df_raw.head())
```
