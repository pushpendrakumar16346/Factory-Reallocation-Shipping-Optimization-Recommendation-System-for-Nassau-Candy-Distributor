```python

# Feature Engineering, Model, Dashboard and Final Project Validation

import ast
import io
import json
import inspect
import hashlib
import importlib.util
import traceback
from pathlib import Path

import numpy as np
import pandas as pd
import joblib


print("=" * 75)
print("FEATURE ENGINEERING AND APPLICATION VALIDATION")
print("=" * 75)


# Prepare validation output

APP_REPORTS.mkdir(parents=True, exist_ok=True)

validation_results = []
detailed_results = []
claims_results = []

def record_check(area, check, status, evidence=""):
    validation_results.append({
        "Validation Area": area,
        "Check": check,
        "Status": status,
        "Evidence": str(evidence)
    })

    print(f"{check}: {status}")
    if evidence:
        print("Evidence:", evidence)


def safe_read_source(path):
    if path.exists():
        return path.read_text(
            encoding="utf-8",
            errors="replace"
        )
    return ""


print("\nFEATURE FORMULA VERIFICATION")
print("-" * 75)

# Recalculate lead time independently from the original date columns.

required_date_columns = ["Order Date", "Ship Date"]

lead_time_columns = [
    col for col in processed_df.columns
    if "lead" in col.lower() and "time" in col.lower()
]

if all(col in processed_df.columns for col in required_date_columns):
    order_dates = pd.to_datetime(
        processed_df["Order Date"],
        dayfirst=True,
        errors="coerce"
    )

    ship_dates = pd.to_datetime(
        processed_df["Ship Date"],
        dayfirst=True,
        errors="coerce"
    )

    calculated_lead_time = (
        ship_dates - order_dates
    ).dt.days

    if lead_time_columns:
        for feature in lead_time_columns:
            stored_lead_time = pd.to_numeric(
                processed_df[feature],
                errors="coerce"
            )

            both_missing = (
                calculated_lead_time.isna()
                & stored_lead_time.isna()
            )

            both_present = (
                calculated_lead_time.notna()
                & stored_lead_time.notna()
            )

            comparable = both_present | both_missing

            matching_values = (
                calculated_lead_time[both_present].to_numpy()
                == stored_lead_time[both_present].to_numpy()
            )

            null_pattern_matches = (
                calculated_lead_time.isna().to_numpy()
                == stored_lead_time.isna().to_numpy()
            ).all()

            formula_matches = (
                matching_values.all()
                and null_pattern_matches
                and comparable.all()
            )

            record_check(
                "Feature Formula QA",
                f"Lead time formula: {feature}",
                "PASS" if formula_matches else "FAIL",
                f"Compared records: {int(both_present.sum())}; "
                f"negative calculated values: "
                f"{int((calculated_lead_time < 0).sum())}"
            )
    else:
        record_check(
            "Feature Formula QA",
            "Lead time feature identification",
            "REVIEW REQUIRED",
            "No lead-time column identified in processed_df."
        )
else:
    record_check(
        "Feature Formula QA",
        "Required date columns",
        "FAIL",
        "Order Date or Ship Date is missing."
    )


# Inspect feature utility source for other engineered formulas.

feature_utils_path = APP_SRC / "feature_utils.py"
feature_source = safe_read_source(feature_utils_path)

if feature_source:
    feature_tree = ast.parse(feature_source)

    feature_functions = [
        node.name
        for node in ast.walk(feature_tree)
        if isinstance(node, (ast.FunctionDef, ast.AsyncFunctionDef))
    ]

    print("\nFeature utility functions:")
    for function_name in feature_functions:
        print("-", function_name)

    record_check(
        "Feature Formula QA",
        "Feature utility source inspection",
        "PASS",
        f"Inspected {feature_utils_path.name}; "
        f"functions found: {len(feature_functions)}"
    )
else:
    record_check(
        "Feature Formula QA",
        "Feature utility source inspection",
        "REVIEW REQUIRED",
        "feature_utils.py was not found."
    )

record_check(
    "Feature Formula QA",
    "All engineered feature formulas",
    "REVIEW REQUIRED",
    "Only independently reproducible formulas can be confirmed. "
    "Review remaining feature definitions individually."
)


print("\nMODEL RELOAD VERIFICATION")
print("-" * 75)

# Reload project model artifacts and record successful or failed loads.

model_files = []

if APP_MODELS.exists():
    model_files = [
        path for path in APP_MODELS.rglob("*")
        if path.is_file()
        and path.suffix.lower() in [
            ".pkl", ".pickle", ".joblib"
        ]
    ]

loaded_models = {}

for model_path in model_files:
    try:
        loaded_model = joblib.load(model_path)
        loaded_models[model_path.name] = loaded_model

        record_check(
            "Model QA",
            f"Reload model: {model_path.name}",
            "PASS",
            f"Loaded object type: {type(loaded_model).__name__}"
        )

    except Exception as error:
        record_check(
            "Model QA",
            f"Reload model: {model_path.name}",
            "FAIL",
            str(error)
        )

if not model_files:
    record_check(
        "Model QA",
        "Model artifact discovery",
        "REVIEW REQUIRED",
        "No .pkl, .pickle or .joblib files found in APP_MODELS."
    )


print("\nPREDICTION REPRODUCIBILITY")
print("-" * 75)

# Use model feature_names_in_ where available.
# Compare predictions from the original loaded model and a fresh reload.

prediction_tested = False

for model_path in model_files:
    if model_path.name not in loaded_models:
        continue

    model = loaded_models[model_path.name]

    if not hasattr(model, "predict"):
        record_check(
            "Reproducibility",
            f"Prediction interface: {model_path.name}",
            "REVIEW REQUIRED",
            "Loaded object does not expose predict()."
        )
        continue

    expected_features = getattr(
        model,
        "feature_names_in_",
        None
    )

    if expected_features is None:
        record_check(
            "Reproducibility",
            f"Prediction test: {model_path.name}",
            "REVIEW REQUIRED",
            "Model does not expose feature_names_in_; "
            "safe input alignment cannot be confirmed."
        )
        continue

    expected_features = list(expected_features)

    missing_features = [
        col for col in expected_features
        if col not in processed_df.columns
    ]

    if missing_features:
        record_check(
            "Reproducibility",
            f"Prediction test: {model_path.name}",
            "REVIEW REQUIRED",
            f"Missing model features: {missing_features}"
        )
        continue

    try:
        sample_input = processed_df[
            expected_features
        ].head(20).copy()

        if sample_input.empty:
            raise ValueError("No sample records available.")

        first_prediction = np.asarray(
            model.predict(sample_input)
        )

        repeated_prediction = np.asarray(
            model.predict(sample_input)
        )

        reloaded_model = joblib.load(model_path)

        reloaded_prediction = np.asarray(
            reloaded_model.predict(sample_input)
        )

        repeated_match = np.allclose(
            first_prediction,
            repeated_prediction,
            equal_nan=True
        )

        reload_match = np.allclose(
            first_prediction,
            reloaded_prediction,
            equal_nan=True
        )

        prediction_tested = True

        record_check(
            "Reproducibility",
            f"Repeated prediction: {model_path.name}",
            "PASS" if repeated_match else "FAIL",
            f"Sample size: {len(sample_input)}"
        )

        record_check(
            "Reproducibility",
            f"Prediction after reload: {model_path.name}",
            "PASS" if reload_match else "FAIL",
            "Compared original and freshly reloaded model outputs."
        )

    except Exception as error:
        record_check(
            "Reproducibility",
            f"Prediction test: {model_path.name}",
            "REVIEW REQUIRED",
            str(error)
        )

if not prediction_tested:
    record_check(
        "Reproducibility",
        "Overall prediction reproducibility",
        "REVIEW REQUIRED",
        "No model could be tested with confirmed feature alignment."
    )


print("\nRECOMMENDATION FORMULA VERIFICATION")
print("-" * 75)

# Inspect recommendation functions and their implementation.

recommendation_path = APP_SRC / "recommendation_utils.py"
recommendation_source = safe_read_source(recommendation_path)

recommendation_functions = []

if recommendation_source:
    recommendation_tree = ast.parse(recommendation_source)

    recommendation_functions = [
        node.name
        for node in ast.walk(recommendation_tree)
        if isinstance(node, (ast.FunctionDef, ast.AsyncFunctionDef))
    ]

    print("Recommendation functions:")
    for function_name in recommendation_functions:
        print("-", function_name)

    print("\nRecommendation source:")
    print(recommendation_source)

    record_check(
        "Score QA",
        "Recommendation source inspection",
        "PASS",
        f"Functions identified: {len(recommendation_functions)}"
    )

    record_check(
        "Score QA",
        "Recommendation formula verification",
        "REVIEW REQUIRED",
        "Source inspected, but expected score values and "
        "independent reference calculations are not established."
    )
else:
    record_check(
        "Score QA",
        "Recommendation utility inspection",
        "REVIEW REQUIRED",
        "recommendation_utils.py not found."
    )


print("\nSLIDER AND FILTER UI VERIFICATION")
print("-" * 75)

# Parse app source and inspect the presence of UI controls.

app_source = safe_read_source(APP_FILE)

if app_source:
    try:
        app_tree = ast.parse(app_source)

        record_check(
            "UI QA",
            "Streamlit application syntax",
            "PASS",
            "Application source parsed successfully."
        )

    except SyntaxError as error:
        record_check(
            "UI QA",
            "Streamlit application syntax",
            "FAIL",
            str(error)
        )
        app_tree = None
else:
    app_tree = None
    record_check(
        "UI QA",
        "Streamlit application source",
        "FAIL",
        "app.py was not found."
    )


# Check whether the app source contains slider controls.

slider_found = (
    "slider(" in app_source
    or "select_slider(" in app_source
)

record_check(
    "UI QA",
    "Slider control presence",
    "PASS" if slider_found else "REVIEW REQUIRED",
    "Static source inspection only; live slider interaction "
    "has not been tested."
)


# Check product, region and ship mode filters independently.

filter_definitions = {
    "Product filter": [
        "product",
        "product_name",
        "product_category"
    ],
    "Region filter": [
        "region",
        "region_name"
    ],
    "Ship mode filter": [
        "ship_mode",
        "ship mode"
    ]
}

for filter_name, keywords in filter_definitions.items():
    found = any(
        keyword.lower() in app_source.lower()
        for keyword in keywords
    )

    record_check(
        "UI QA",
        f"{filter_name} source presence",
        "PASS" if found else "REVIEW REQUIRED",
        "Static source check; actual filter behavior "
        "requires live application testing."
    )


print("\nSCENARIO TESTING")
print("-" * 75)

# Inspect scenario functions and their input definitions.

scenario_path = APP_SRC / "scenario_utils.py"
scenario_source = safe_read_source(scenario_path)

if scenario_source:
    scenario_tree = ast.parse(scenario_source)

    scenario_functions = [
        node.name
        for node in ast.walk(scenario_tree)
        if isinstance(node, (ast.FunctionDef, ast.AsyncFunctionDef))
    ]

    print("Scenario functions:")
    for function_name in scenario_functions:
        print("-", function_name)

    print("\nScenario utility source:")
    print(scenario_source)

    record_check(
        "Scenario QA",
        "Scenario utility source inspection",
        "PASS",
        f"Functions identified: {len(scenario_functions)}"
    )

    record_check(
        "Scenario QA",
        "Scenario output verification",
        "REVIEW REQUIRED",
        "Expected scenario outputs and independent benchmark "
        "values must be confirmed before numerical validation."
    )
else:
    record_check(
        "Scenario QA",
        "Scenario utility inspection",
        "REVIEW REQUIRED",
        "scenario_utils.py not found."
    )


print("\nDOWNLOAD AND EXPORT TESTING")
print("-" * 75)

# Generate CSV and Excel files in memory, then read them back.
# This verifies file generation and data readability.

export_results = []

try:
    csv_buffer = io.StringIO()
    processed_df.to_csv(csv_buffer, index=False)

    csv_content = csv_buffer.getvalue()
    csv_check = pd.read_csv(io.StringIO(csv_content))

    csv_valid = (
        csv_check.shape == processed_df.shape
        and list(csv_check.columns) == list(processed_df.columns)
    )

    record_check(
        "Export QA",
        "CSV export and read-back",
        "PASS" if csv_valid else "FAIL",
        f"Exported shape: {csv_check.shape}"
    )

    export_results.append({
        "Export": "CSV",
        "Status": "PASS" if csv_valid else "FAIL",
        "Rows": len(csv_check),
        "Columns": len(csv_check.columns)
    })

except Exception as error:
    record_check(
        "Export QA",
        "CSV export and read-back",
        "FAIL",
        str(error)
    )


try:
    excel_buffer = io.BytesIO()

    with pd.ExcelWriter(
        excel_buffer,
        engine="openpyxl"
    ) as writer:
        processed_df.to_excel(
            writer,
            sheet_name="Processed Data",
            index=False
        )

    excel_bytes = excel_buffer.getvalue()

    excel_check = pd.read_excel(
        io.BytesIO(excel_bytes),
        sheet_name="Processed Data"
    )

    excel_valid = (
        excel_check.shape == processed_df.shape
        and list(excel_check.columns) == list(processed_df.columns)
    )

    record_check(
        "Export QA",
        "Excel export and read-back",
        "PASS" if excel_valid else "FAIL",
        f"Exported shape: {excel_check.shape}; "
        f"file size: {len(excel_bytes)} bytes"
    )

    export_results.append({
        "Export": "Excel",
        "Status": "PASS" if excel_valid else "FAIL",
        "Rows": len(excel_check),
        "Columns": len(excel_check.columns)
    })

except Exception as error:
    record_check(
        "Export QA",
        "Excel export and read-back",
        "FAIL",
        str(error)
    )

export_report = pd.DataFrame(export_results)
display(export_report)


print("\nEND-TO-END APPLICATION VALIDATION")
print("-" * 75)

# Confirm that the main app, configuration and data bundle exist.

record_check(
    "End-to-End QA",
    "Application file availability",
    "PASS" if APP_FILE.exists() else "FAIL",
    str(APP_FILE)
)

record_check(
    "End-to-End QA",
    "Application data directory",
    "PASS" if APP_DATA.exists() else "FAIL",
    str(APP_DATA)
)

record_check(
    "End-to-End QA",
    "Application model directory",
    "PASS" if APP_MODELS.exists() else "REVIEW REQUIRED",
    str(APP_MODELS)
)

config_path = APP_ROOT / "app_config.json"

if config_path.exists():
    try:
        with open(config_path, "r", encoding="utf-8") as file:
            json.load(file)

        config_status = "PASS"
        config_evidence = "Configuration JSON parsed successfully."

    except Exception as error:
        config_status = "FAIL"
        config_evidence = str(error)
else:
    config_status = "REVIEW REQUIRED"
    config_evidence = "app_config.json not found."

record_check(
    "End-to-End QA",
    "Application configuration",
    config_status,
    config_evidence
)


# Static checks cannot prove that the app works in a real browser.
# Keep interactive tests explicitly marked as not tested.

manual_browser_checks = [
    "Application opens in browser",
    "All dashboard pages load",
    "Slider changes update scenario outputs",
    "Product filter updates displayed data",
    "Region filter updates displayed data",
    "Ship mode filter updates displayed data",
    "Scenario outputs respond correctly",
    "CSV download works from dashboard",
    "Excel download works from dashboard"
]

browser_report = pd.DataFrame({
    "Browser Test": manual_browser_checks,
    "Status": ["NOT TESTED"] * len(manual_browser_checks),
    "Evidence": [
        "Requires live browser interaction."
    ] * len(manual_browser_checks)
})

display(browser_report)

record_check(
    "End-to-End QA",
    "Live browser end-to-end testing",
    "NOT TESTED",
    "Static validation completed; interactive browser evidence "
    "must be recorded separately."
)


print("\nCLAIMS AND LIMITATIONS AUDIT")
print("-" * 75)

# Preserve the project's documented analytical limitations.

claims_and_limitations = [
    {
        "Area": "Lead Time",
        "Claim or Limitation": (
            "Lead time is calculated as Ship Date minus Order Date. "
            "It is not independently verified carrier transit time."
        )
    },
    {
        "Area": "Factory Mapping",
        "Claim or Limitation": (
            "Product-to-factory mapping is a reference mapping, "
            "not verified historical dispatch."
        )
    },
    {
        "Area": "Model Predictors",
        "Claim or Limitation": (
            "The model uses 20 predictor features. Factory is not "
            "included as a model predictor."
        )
    },
    {
        "Area": "Geographic Distance",
        "Claim or Limitation": (
            "Haversine distances represent straight-line distance, "
            "not road or freight-route distance."
        )
    },
    {
        "Area": "Cost Savings",
        "Claim or Limitation": (
            "Alternative factory freight or manufacturing cost "
            "savings have not been independently established."
        )
    },
    {
        "Area": "Scenario Analysis",
        "Claim or Limitation": (
            "Scenario support is not a probability of success."
        )
    },
    {
        "Area": "Recommendations",
        "Claim or Limitation": (
            "Recommendations are indicative and are not proven "
            "to be globally optimal."
        )
    },
    {
        "Area": "Model Metrics",
        "Claim or Limitation": (
            "Existing model metric discrepancies must remain "
            "documented and must not be silently reconciled."
        )
    }
]

claims_report = pd.DataFrame(claims_and_limitations)
display(claims_report)

record_check(
    "Evidence Audit",
    "Project claims and limitations documented",
    "PASS",
    f"Documented limitations: {len(claims_report)}"
)


print("\nFINAL VALIDATION REPORT")
print("-" * 75)

validation_report = pd.DataFrame(validation_results)

display(validation_report)

status_summary = (
    validation_report["Status"]
    .value_counts()
    .rename_axis("Status")
    .reset_index(name="Count")
)

display(status_summary)


print("\nSAVE VALIDATION EVIDENCE")
print("-" * 75)

validation_report_path = (
    APP_REPORTS / "application_validation_report.csv"
)

claims_report_path = (
    APP_REPORTS / "claims_and_limitations_audit.csv"
)

browser_report_path = (
    APP_REPORTS / "browser_testing_checklist.csv"
)

export_report_path = (
    APP_REPORTS / "export_validation_report.csv"
)

validation_report.to_csv(
    validation_report_path,
    index=False
)

claims_report.to_csv(
    claims_report_path,
    index=False
)

browser_report.to_csv(
    browser_report_path,
    index=False
)

export_report.to_csv(
    export_report_path,
    index=False
)

print("Validation report:", validation_report_path)
print("Claims and limitations:", claims_report_path)
print("Browser checklist:", browser_report_path)
print("Export report:", export_report_path)


print("\n" + "=" * 75)
print("VALIDATION EXECUTION COMPLETED")
print("=" * 75)

print("PASS:",
      (validation_report["Status"] == "PASS").sum())

print("FAIL:",
      (validation_report["Status"] == "FAIL").sum())

print("REVIEW REQUIRED:",
      (validation_report["Status"] == "REVIEW REQUIRED").sum())

print("NOT TESTED:",
      (validation_report["Status"] == "NOT TESTED").sum())

print("\nReports saved successfully.")
print("Review all FAIL and REVIEW REQUIRED results.")
print("Complete the browser checklist after testing the live dashboard.")
```
