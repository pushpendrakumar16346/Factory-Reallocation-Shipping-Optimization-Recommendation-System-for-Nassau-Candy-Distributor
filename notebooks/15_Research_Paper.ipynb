```python

from pathlib import Path
import json
import re
import sys
import importlib.util

# ---------------------------------------------------------
# Set project paths
# ---------------------------------------------------------

PROJECT_ROOT = Path(r"C:\Users\hp\Nassau_Candy_Final_Submission")
NOTEBOOK_DIR = PROJECT_ROOT / "notebooks"
RESEARCH_DIR = PROJECT_ROOT / "reports" / "research_paper"

RESEARCH_DIR.mkdir(parents=True, exist_ok=True)

PAPER_PATH = RESEARCH_DIR / "research_paper_draft.md"

if not PAPER_PATH.exists():
    raise FileNotFoundError(
        f"Research paper draft not found: {PAPER_PATH}"
    )

paper_text = PAPER_PATH.read_text(encoding="utf-8")


# ---------------------------------------------------------
# Search saved notebook outputs for supporting evidence
# ---------------------------------------------------------

search_terms = [
    "MAE", "RMSE", "MSE", "R-squared", "R2",
    "RandomForest", "GradientBoosting", "LinearRegression",
    "product performance", "regional performance",
    "ship mode", "lead time", "profitability",
    "cluster", "revenue", "sales", "silhouette"
]

evidence = []

for notebook_path in sorted(NOTEBOOK_DIR.glob("*.ipynb")):
    try:
        notebook = json.loads(
            notebook_path.read_text(encoding="utf-8")
        )

        for cell_number, cell in enumerate(
            notebook.get("cells", []), start=1
        ):
            for output in cell.get("outputs", []):
                output_text = ""

                if "text" in output:
                    output_text += "".join(output["text"])

                if "data" in output:
                    data = output["data"]
                    output_text += "".join(
                        data.get("text/plain", [])
                    )

                if not output_text.strip():
                    continue

                matched_terms = [
                    term for term in search_terms
                    if term.lower() in output_text.lower()
                ]

                if matched_terms:
                    evidence.append({
                        "notebook": notebook_path.name,
                        "cell": cell_number,
                        "terms": matched_terms,
                        "output": output_text[:5000]
                    })

    except Exception as error:
        print(
            "Notebook inspection warning:",
            notebook_path.name,
            str(error)
        )


# ---------------------------------------------------------
# Save evidence report for factual verification
# ---------------------------------------------------------

evidence_path = RESEARCH_DIR / "notebook_evidence_report.txt"

with evidence_path.open("w", encoding="utf-8") as file:
    file.write("RESEARCH PAPER - NOTEBOOK EVIDENCE REPORT\n")
    file.write("=" * 65 + "\n\n")

    if evidence:
        for item in evidence:
            file.write("NOTEBOOK: " + item["notebook"] + "\n")
            file.write("CELL: " + str(item["cell"]) + "\n")
            file.write(
                "MATCHED TERMS: "
                + ", ".join(item["terms"])
                + "\n"
            )
            file.write("OUTPUT:\n")
            file.write(item["output"] + "\n")
            file.write("-" * 65 + "\n\n")
    else:
        file.write(
            "No matching saved outputs were found.\n"
            "Run the relevant analysis notebooks and save their outputs.\n"
        )


# ---------------------------------------------------------
# Identify incomplete content
# ---------------------------------------------------------

review_patterns = [
    r"\[INSERT[^\]]*\]",
    r"\[.*?VERIFIED.*?\]",
    r"to be added",
    r"placeholder"
]

incomplete_lines = []

for line_number, line in enumerate(
    paper_text.splitlines(), start=1
):
    if any(
        re.search(pattern, line, flags=re.IGNORECASE)
        for pattern in review_patterns
    ):
        incomplete_lines.append((line_number, line.strip()))


# ---------------------------------------------------------
# Review references section
# ---------------------------------------------------------

references_match = re.search(
    r"# References\s*(.*?)(?=\n# |\Z)",
    paper_text,
    flags=re.IGNORECASE | re.DOTALL
)

references_text = (
    references_match.group(1).strip()
    if references_match else ""
)

reference_count = len(
    re.findall(r"(?m)^\s*\d+\.", references_text)
)

references_present = reference_count > 0


# ---------------------------------------------------------
# Generate Word document
# ---------------------------------------------------------

word_path = RESEARCH_DIR / "research_paper_draft.docx"
word_status = "NOT CREATED"

if importlib.util.find_spec("docx") is not None:
    from docx import Document
    from docx.shared import Inches, Pt
    from docx.enum.text import WD_ALIGN_PARAGRAPH

    document = Document()

    section = document.sections[0]
    section.top_margin = Inches(0.8)
    section.bottom_margin = Inches(0.8)
    section.left_margin = Inches(1)
    section.right_margin = Inches(1)

    styles = document.styles
    styles["Normal"].font.name = "Times New Roman"
    styles["Normal"].font.size = Pt(12)

    styles["Normal"].paragraph_format.line_spacing = 1.5

    lines = paper_text.splitlines()
    first_heading = True

    for line in lines:
        stripped = line.strip()

        if not stripped or stripped == "---":
            continue

        if stripped.startswith("# "):
            heading = stripped[2:].strip()

            if first_heading:
                paragraph = document.add_paragraph()
                paragraph.alignment = WD_ALIGN_PARAGRAPH.CENTER
                run = paragraph.add_run(heading)
                run.bold = True
                run.font.size = Pt(16)
                first_heading = False
            else:
                document.add_heading(heading, level=1)

        elif stripped.startswith("## "):
            document.add_heading(stripped[3:].strip(), level=2)

        elif stripped.startswith("- ") or stripped.startswith("[ ] "):
            content = stripped[2:] if stripped.startswith("- ") else stripped[4:]
            paragraph = document.add_paragraph(
                style="List Bullet"
            )
            paragraph.add_run(content)

        else:
            paragraph = document.add_paragraph()
            paragraph.add_run(
                stripped.replace("**", "").replace("*", "")
            )

    document.save(word_path)
    word_status = "CREATED"
else:
    word_status = "python-docx not installed"


# ---------------------------------------------------------
# Generate PDF document
# ---------------------------------------------------------

pdf_path = RESEARCH_DIR / "research_paper_draft.pdf"
pdf_status = "NOT CREATED"

if importlib.util.find_spec("reportlab") is not None:
    from reportlab.lib.pagesizes import A4
    from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle
    from reportlab.lib.enums import TA_CENTER
    from reportlab.lib.units import inch
    from reportlab.platypus import (
        SimpleDocTemplate,
        Paragraph,
        Spacer,
        PageBreak
    )
    from xml.sax.saxutils import escape

    pdf_document = SimpleDocTemplate(
        str(pdf_path),
        pagesize=A4,
        rightMargin=0.75 * inch,
        leftMargin=0.75 * inch,
        topMargin=0.75 * inch,
        bottomMargin=0.75 * inch
    )

    styles = getSampleStyleSheet()

    title_style = ParagraphStyle(
        "PaperTitle",
        parent=styles["Title"],
        alignment=TA_CENTER,
        fontName="Times-Bold",
        fontSize=16,
        leading=21,
        spaceAfter=18
    )

    heading_style = ParagraphStyle(
        "PaperHeading",
        parent=styles["Heading1"],
        fontName="Times-Bold",
        fontSize=14,
        leading=18,
        spaceBefore=14,
        spaceAfter=8
    )

    body_style = ParagraphStyle(
        "PaperBody",
        parent=styles["BodyText"],
        fontName="Times-Roman",
        fontSize=11,
        leading=17,
        spaceAfter=8
    )

    story = []

    for line in paper_text.splitlines():
        stripped = line.strip()

        if not stripped or stripped == "---":
            continue

        if stripped.startswith("# "):
            heading = escape(stripped[2:].strip())

            if not story:
                story.append(Paragraph(heading, title_style))
            else:
                story.append(Spacer(1, 8))
                story.append(Paragraph(heading, heading_style))

        elif stripped.startswith("## "):
            story.append(
                Paragraph(
                    escape(stripped[3:].strip()),
                    heading_style
                )
            )

        else:
            content = escape(
                stripped.replace("**", "").replace("*", "")
            )
            story.append(Paragraph(content, body_style))

    pdf_document.build(story)
    pdf_status = "CREATED"
else:
    pdf_status = "reportlab not installed"


# ---------------------------------------------------------
# Final audit report
# ---------------------------------------------------------

audit_path = RESEARCH_DIR / "final_audit_report.txt"

if incomplete_lines:
    content_status = "REVIEW REQUIRED"
else:
    content_status = "NO KNOWN PLACEHOLDERS FOUND"

if references_present:
    reference_status = (
        f"{reference_count} numbered reference entries found; "
        "source verification still required."
    )
else:
    reference_status = "REVIEW REQUIRED: References section missing or empty."

with audit_path.open("w", encoding="utf-8") as file:
    file.write("FINAL RESEARCH PAPER AUDIT\n")
    file.write("=" * 65 + "\n\n")

    file.write("Content status: " + content_status + "\n")
    file.write("Reference status: " + reference_status + "\n")
    file.write("Notebook evidence items: " + str(len(evidence)) + "\n")
    file.write("Incomplete content items: " + str(len(incomplete_lines)) + "\n\n")

    if incomplete_lines:
        file.write("ITEMS REQUIRING REVIEW:\n")
        for line_number, line in incomplete_lines:
            file.write(f"Line {line_number}: {line}\n")
    else:
        file.write("No known placeholder terms detected.\n")

    file.write("\nOUTPUT FILES:\n")
    file.write(str(PAPER_PATH) + "\n")
    file.write(str(evidence_path) + "\n")
    file.write(str(word_path) + " - " + word_status + "\n")
    file.write(str(pdf_path) + " - " + pdf_status + "\n")


# ---------------------------------------------------------
# Display completion summary
# ---------------------------------------------------------

print("=" * 65)
print("RESEARCH PAPER FINAL PREPARATION")
print("=" * 65)

print("\nNotebook evidence items:", len(evidence))
print("Incomplete content items:", len(incomplete_lines))
print("References:", reference_status)

print("\nWord document:", word_status)
print("PDF document:", pdf_status)

print("\nFiles saved:")
print("Research paper:", PAPER_PATH)
print("Evidence report:", evidence_path)
print("Word document:", word_path if word_status == "CREATED" else word_status)
print("PDF document:", pdf_path if pdf_status == "CREATED" else pdf_status)
print("Final audit:", audit_path)

print("\nContent status:", content_status)

if incomplete_lines:
    print("\nREVIEW REQUIRED ITEMS:")
    for line_number, line in incomplete_lines:
        print(f"Line {line_number}: {line}")

print("\n" + "=" * 65)

if incomplete_lines or not references_present:
    print("STATUS: DRAFT GENERATED - FINAL VERIFICATION PENDING")
else:
    print("STATUS: NO KNOWN PLACEHOLDERS - HUMAN REVIEW STILL REQUIRED")

print("=" * 65)
```
