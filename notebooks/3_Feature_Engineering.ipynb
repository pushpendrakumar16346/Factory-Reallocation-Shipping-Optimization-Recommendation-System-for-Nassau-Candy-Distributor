```python
# LOAD DATA AND CREATE SAFE FEATURE ENGINEERING COPY

import pandas as pd
import numpy as np
from pathlib import Path

PROJECT_ROOT = Path(r"C:\Users\hp\Nassau_Candy_Final_Submission")

RAW_CSV_PATH = PROJECT_ROOT / "data" / "raw" / "Nassau Candy Distributor.csv"
PROCESSED_FOLDER = PROJECT_ROOT / "data" / "processed"

PROCESSED_FOLDER.mkdir(parents=True, exist_ok=True)

df_raw = pd.read_csv(RAW_CSV_PATH)

# Independent working copy
df_feature = df_raw.copy(deep=True)

# Parse dates into separate helper columns
df_feature["Order_Date_Parsed"] = pd.to_datetime(
    df_raw["Order Date"], dayfirst=True, errors="coerce"
)

df_feature["Ship_Date_Parsed"] = pd.to_datetime(
    df_raw["Ship Date"], dayfirst=True, errors="coerce"
)

print("Dataset loaded:", df_feature.shape)
print("Order dates parsed:", df_feature["Order_Date_Parsed"].notna().sum())
print("Ship dates parsed:", df_feature["Ship_Date_Parsed"].notna().sum())

display(df_feature.head())
```

```python
# CALCULATE SHIPPING LEAD TIME IN DAYS

df_feature["Lead_Time_Days"] = (
    df_feature["Ship_Date_Parsed"] -
    df_feature["Order_Date_Parsed"]
).dt.days

print("Lead Time Feature Created")

display(
    df_feature[[
        "Order Date",
        "Ship Date",
        "Lead_Time_Days"
    ]].head(10)
)

display(df_feature["Lead_Time_Days"].describe())
```

```python
# EXTRACT ORDER YEAR

df_feature["Order_Year"] = (
    df_feature["Order_Date_Parsed"].dt.year
)

display(df_feature[["Order Date", "Order_Year"]].head())
```

```python
# EXTRACT ORDER MONTH

df_feature["Order_Month"] = (
    df_feature["Order_Date_Parsed"].dt.month
)

display(df_feature[["Order Date", "Order_Month"]].head())
```

```python
# EXTRACT ORDER QUARTER

df_feature["Order_Quarter"] = (
    df_feature["Order_Date_Parsed"].dt.quarter
)

display(df_feature[["Order Date", "Order_Quarter"]].head())
```

```python
# EXTRACT ORDER WEEKDAY

df_feature["Order_Weekday"] = (
    df_feature["Order_Date_Parsed"].dt.day_name()
)

display(df_feature[["Order Date", "Order_Weekday"]].head())
```

```python
# EXTRACT SHIP YEAR

df_feature["Ship_Year"] = (
    df_feature["Ship_Date_Parsed"].dt.year
)

display(df_feature[["Ship Date", "Ship_Year"]].head())
```

```python
# EXTRACT SHIP MONTH

df_feature["Ship_Month"] = (
    df_feature["Ship_Date_Parsed"].dt.month
)

display(df_feature[["Ship Date", "Ship_Month"]].head())
```

```python
# CALCULATE PROFIT MARGIN PERCENTAGE

df_feature["Profit_Margin_Pct"] = np.where(
    df_feature["Sales"] != 0,
    (df_feature["Gross Profit"] / df_feature["Sales"]) * 100,
    np.nan
)

display(
    df_feature[[
        "Sales",
        "Gross Profit",
        "Profit_Margin_Pct"
    ]].head(10)
)

print(
    "Average Profit Margin (%):",
    round(df_feature["Profit_Margin_Pct"].mean(), 2)
)
```

```python
# CALCULATE SALES PER UNIT

df_feature["Sales_Per_Unit"] = np.where(
    df_feature["Units"] > 0,
    df_feature["Sales"] / df_feature["Units"],
    np.nan
)

display(
    df_feature[["Sales", "Units", "Sales_Per_Unit"]].head(10)
)
```

```python
# CALCULATE COST PER UNIT

df_feature["Cost_Per_Unit"] = np.where(
    df_feature["Units"] > 0,
    df_feature["Cost"] / df_feature["Units"],
    np.nan
)

display(
    df_feature[["Cost", "Units", "Cost_Per_Unit"]].head(10)
)
```

```python
# CALCULATE PROFIT PER UNIT

df_feature["Profit_Per_Unit"] = np.where(
    df_feature["Units"] > 0,
    df_feature["Gross Profit"] / df_feature["Units"],
    np.nan
)

display(
    df_feature[[
        "Gross Profit",
        "Units",
        "Profit_Per_Unit"
    ]].head(10)
)
```

```python
# CREATE LEAD TIME CATEGORIES USING DATA DISTRIBUTION

lead_time_values = df_feature["Lead_Time_Days"]

lead_q1 = lead_time_values.quantile(0.33)
lead_q2 = lead_time_values.quantile(0.67)

df_feature["Lead_Time_Category"] = np.select(
    [
        lead_time_values <= lead_q1,
        (lead_time_values > lead_q1) & (lead_time_values <= lead_q2),
        lead_time_values > lead_q2
    ],
    [
        "Low",
        "Medium",
        "High"
    ],
    default="Unknown"
)

print("Lead Time Category Thresholds:")
print("Low:", lead_q1)
print("Medium:", lead_q2)

display(df_feature["Lead_Time_Category"].value_counts(dropna=False))
```

```python
# CREATE PROFIT CATEGORIES USING GROSS PROFIT DISTRIBUTION

profit_q1 = df_feature["Gross Profit"].quantile(0.33)
profit_q2 = df_feature["Gross Profit"].quantile(0.67)

df_feature["Profit_Category"] = np.select(
    [
        df_feature["Gross Profit"] <= profit_q1,
        (
            (df_feature["Gross Profit"] > profit_q1) &
            (df_feature["Gross Profit"] <= profit_q2)
        ),
        df_feature["Gross Profit"] > profit_q2
    ],
    [
        "Low",
        "Medium",
        "High"
    ],
    default="Unknown"
)

print("Profit Category Thresholds:")
print("Low:", profit_q1)
print("Medium:", profit_q2)

display(df_feature["Profit_Category"].value_counts(dropna=False))
```

```python
# CREATE SALES CATEGORIES USING SALES DISTRIBUTION

sales_q1 = df_feature["Sales"].quantile(0.33)
sales_q2 = df_feature["Sales"].quantile(0.67)

df_feature["Sales_Category"] = np.select(
    [
        df_feature["Sales"] <= sales_q1,
        (
            (df_feature["Sales"] > sales_q1) &
            (df_feature["Sales"] <= sales_q2)
        ),
        df_feature["Sales"] > sales_q2
    ],
    [
        "Low",
        "Medium",
        "High"
    ],
    default="Unknown"
)

print("Sales Category Thresholds:")
print("Low:", sales_q1)
print("Medium:", sales_q2)

display(df_feature["Sales_Category"].value_counts(dropna=False))
```

```python
# VALIDATE ALL DERIVED FEATURES AND FORMULAS

expected_lead_time = (
    df_feature["Ship_Date_Parsed"] -
    df_feature["Order_Date_Parsed"]
).dt.days

expected_profit_margin = np.where(
    df_feature["Sales"] != 0,
    df_feature["Gross Profit"] / df_feature["Sales"] * 100,
    np.nan
)

expected_sales_per_unit = np.where(
    df_feature["Units"] > 0,
    df_feature["Sales"] / df_feature["Units"],
    np.nan
)

expected_cost_per_unit = np.where(
    df_feature["Units"] > 0,
    df_feature["Cost"] / df_feature["Units"],
    np.nan
)

expected_profit_per_unit = np.where(
    df_feature["Units"] > 0,
    df_feature["Gross Profit"] / df_feature["Units"],
    np.nan
)

feature_validation = pd.DataFrame({
    "Feature": [
        "Lead_Time_Days",
        "Order_Year",
        "Order_Month",
        "Order_Quarter",
        "Order_Weekday",
        "Ship_Year",
        "Ship_Month",
        "Profit_Margin_Pct",
        "Sales_Per_Unit",
        "Cost_Per_Unit",
        "Profit_Per_Unit",
        "Lead_Time_Category",
        "Profit_Category",
        "Sales_Category"
    ],
    "Missing Values": [
        df_feature[col].isna().sum()
        for col in [
            "Lead_Time_Days",
            "Order_Year",
            "Order_Month",
            "Order_Quarter",
            "Order_Weekday",
            "Ship_Year",
            "Ship_Month",
            "Profit_Margin_Pct",
            "Sales_Per_Unit",
            "Cost_Per_Unit",
            "Profit_Per_Unit",
            "Lead_Time_Category",
            "Profit_Category",
            "Sales_Category"
        ]
    ]
})

display(feature_validation)

formula_checks = {
    "Lead Time Formula": np.allclose(
        df_feature["Lead_Time_Days"].to_numpy(dtype=float),
        expected_lead_time.to_numpy(dtype=float),
        equal_nan=True
    ),
    "Profit Margin Formula": np.allclose(
        df_feature["Profit_Margin_Pct"].to_numpy(dtype=float),
        expected_profit_margin,
        equal_nan=True
    ),
    "Sales Per Unit Formula": np.allclose(
        df_feature["Sales_Per_Unit"].to_numpy(dtype=float),
        expected_sales_per_unit,
        equal_nan=True
    ),
    "Cost Per Unit Formula": np.allclose(
        df_feature["Cost_Per_Unit"].to_numpy(dtype=float),
        expected_cost_per_unit,
        equal_nan=True
    ),
    "Profit Per Unit Formula": np.allclose(
        df_feature["Profit_Per_Unit"].to_numpy(dtype=float),
        expected_profit_per_unit,
        equal_nan=True
    )
}

print("\nFORMULA VALIDATION")
for check, result in formula_checks.items():
    print(check, ":", "PASS" if result else "FAIL")

assert all(formula_checks.values()), "One or more feature formulas failed."
```

```python
# Diagnose Numeric Feature Compatibility and Recheck Model Predictions

from pathlib import Path
import sys
import json
import numpy as np
import pandas as pd
import joblib

PROJECT_ROOT = Path(r"C:\Users\hp\Nassau_Candy_Final_Submission")
APP_ROOT = PROJECT_ROOT / "data" / "processed" / "app_bundle"

sys.path.insert(0, str(APP_ROOT))

from src.feature_utils import prepare_model_features

DATA_FILE = APP_ROOT / "data" / "processed_dataset.csv"
MODEL_FILE = APP_ROOT / "models" / "Nassau_Lead_Time_Model.joblib"
SCHEMA_FILE = APP_ROOT / "models" / "feature_schema.json"

data = pd.read_csv(DATA_FILE, low_memory=False)

with open(SCHEMA_FILE, "r", encoding="utf-8") as file:
    schema = json.load(file)

features = schema["predictor_features"]

sample = data.head(100).copy()
transformed = prepare_model_features(sample)

comparison = []

for column in features:
    left = transformed[column].reset_index(drop=True)
    right = sample[column].reset_index(drop=True)

    if pd.api.types.is_numeric_dtype(left) and pd.api.types.is_numeric_dtype(right):
        left_num = pd.to_numeric(left, errors="coerce").to_numpy(dtype=float)
        right_num = pd.to_numeric(right, errors="coerce").to_numpy(dtype=float)

        match = np.allclose(
            left_num,
            right_num,
            rtol=1e-9,
            atol=1e-8,
            equal_nan=True
        )

        valid = np.isfinite(left_num) & np.isfinite(right_num)

        if valid.any():
            max_difference = float(
                np.max(np.abs(left_num[valid] - right_num[valid]))
            )
        else:
            max_difference = 0.0

    else:
        match = left.fillna("<NA>").astype(str).equals(
            right.fillna("<NA>").astype(str)
        )
        max_difference = None

    comparison.append({
        "Feature": column,
        "Processed_Dtype": str(right.dtype),
        "Transformed_Dtype": str(left.dtype),
        "Match": bool(match),
        "Maximum_Absolute_Difference": max_difference
    })

comparison_df = pd.DataFrame(comparison)

print("=" * 90)
print("FEATURE COMPATIBILITY DIAGNOSTIC")
print("=" * 90)

display(comparison_df)

matching_count = int(comparison_df["Match"].sum())
print(f"\nMatching features: {matching_count}/{len(features)}")

print("\nFEATURES REQUIRING REVIEW")
display(comparison_df.loc[~comparison_df["Match"]])

print("\nMODEL PREDICTION VERIFICATION")

model = joblib.load(MODEL_FILE)

X = transformed.loc[:, features]
predictions = model.predict(X)

print("Prediction successful:", len(predictions) == len(sample))
print("Prediction count:", len(predictions))
print("First five predictions:", predictions[:5])

print("\n" + "=" * 90)

if matching_count == len(features) and len(predictions) == len(sample):
    print("FEATURE COMPATIBILITY: PASSED")
    print("MODEL PREDICTION: PASSED")
else:
    print("FEATURE COMPATIBILITY: NEEDS REVIEW")
    print("Do not proceed to final app integration yet.")

print("=" * 90)
```
