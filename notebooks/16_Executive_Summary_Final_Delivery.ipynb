```python

from pathlib import Path
from datetime import datetime
import json
import zipfile
import importlib.util
import ast
import textwrap

# ============================================================
# PROJECT PATHS AND DELIVERY FOLDERS
# ============================================================
PROJECT_ROOT = Path(r"C:\Users\hp\Nassau_Candy_Final_Submission")

DATA_DIR = PROJECT_ROOT / "data"
RAW_DIR = DATA_DIR / "raw"
APP_ROOT = DATA_DIR / "processed" / "app_bundle"
APP_FILE = APP_ROOT / "app.py"

DELIVERY_DIR = PROJECT_ROOT / "reports" / "final_delivery"
SUMMARY_DIR = DELIVERY_DIR / "executive_summary"
PRESENTATION_DIR = DELIVERY_DIR / "presentation"
DOCUMENTATION_DIR = DELIVERY_DIR / "documentation"
VALIDATION_DIR = DELIVERY_DIR / "validation"
PACKAGE_DIR = DELIVERY_DIR / "package"

for folder in [
    SUMMARY_DIR, PRESENTATION_DIR, DOCUMENTATION_DIR,
    VALIDATION_DIR, PACKAGE_DIR
]:
    folder.mkdir(parents=True, exist_ok=True)

print("=" * 70)
print("NASSAU CANDY - FINAL DELIVERY DOCUMENTS")
print("=" * 70)


# ============================================================
# EXECUTIVE SUMMARY AND BUSINESS FINDINGS
# ============================================================
executive_summary = """
# Executive Summary
## Nassau Candy Distributor Analytics Project

### Project Overview

This project analyses historical Nassau Candy Distributor data using
data cleaning, exploratory analysis, feature engineering, predictive
modelling, clustering, factory-reference analysis, scenario simulation
and decision-support methods.

An interactive Streamlit dashboard was developed to make the findings
accessible to business users.

### Business Problem

The project investigates product performance, regional sales,
shipping patterns, profitability and date-derived lead-time patterns.
It also examines whether factory-reference information can support
further operational investigation.

### Dataset

The original dataset contains 10,194 records and 18 columns.
The processed dataset contains 10,194 records and 34 columns.

### Main Findings

- Total sales: $141,783.63.
- Total cost: $48,340.83.
- Gross profit: $93,442.80.
- Weighted gross-profit margin: approximately 65.91%.
- Highest-sales product: Wonka Bar - Triple Dazzle Caramel,
  with sales of $28,485.00.
- Highest-sales region: Pacific, with sales of $46,301.53.
- Lowest-sales region: Gulf, with sales of $22,247.26.
- Largest shipping category: Standard Class, with 6,120 records.

### Predictive Model

The selected algorithm was a Tuned Random Forest Regressor.

Saved model metadata reports:
- MAE: 95.9795
- RMSE: 120.7382
- R-squared: 0.7929
- Test records: 2,062

The model comparison output reports:
- MAE: 139.6707
- RMSE: 163.6754
- R-squared: 0.6195

The chronological holdout output reports:
- MAE: 179.0003
- RMSE: 195.6296
- R-squared: -0.1619

These results come from different saved evaluation outputs and have
not been reconciled. They must not be presented as one consistent
evaluation result. MSE was not separately reported.

### Recommended Actions

1. Review high-sales products and their gross-profit contribution.
2. Investigate regional performance differences.
3. Review shipping performance using verified delivery data.
4. Validate product-to-factory mappings against actual dispatch records.
5. Collect factory capacity, production and cost information.
6. Reconcile model evaluation metrics before operational use.
7. Use the dashboard as a decision-support tool.

### Risks and Limitations

Lead time is calculated as Ship Date minus Order Date and is not
independently verified carrier transit time.

Factory mappings are reference mappings, not confirmed historical
dispatch records. Straight-line distances are not actual freight
routes. Factory-specific savings and scenario success probabilities
have not been established.

Recommendations are indicative, not proven optimal.

### Additional Data Required

- Actual carrier pickup and delivery timestamps.
- Factory production and dispatch records.
- Factory capacity and utilization.
- Factory-specific manufacturing costs.
- Freight rates and practical route distances.
- Inventory and stockout records.
- Customer service-level and delivery performance data.
- Model run logs and evaluation split definitions.

### Conclusion

The project provides an analytical foundation for understanding
product, regional, shipping, financial and factory-reference patterns.

The dashboard makes these findings accessible through an interactive
interface. Further operational data validation is required before
making high-impact factory-allocation, freight-savings or
service-level decisions.
"""

summary_file = SUMMARY_DIR / "executive_summary.md"
summary_file.write_text(
    textwrap.dedent(executive_summary).strip(),
    encoding="utf-8"
)

business_findings = """
# Business Findings

## Financial Performance

| Measure | Result |
|---|---:|
| Total sales | $141,783.63 |
| Total cost | $48,340.83 |
| Gross profit | $93,442.80 |
| Weighted gross-profit margin | 65.9052% |

## Product Performance

The highest-sales product was Wonka Bar - Triple Dazzle Caramel,
with sales of $28,485.00.

## Regional Performance

Pacific recorded the highest sales at $46,301.53 and gross profit
of $30,485.94 across 3,253 records.

Gulf recorded the lowest regional sales at $22,247.26.

## Shipping Performance

Standard Class was the largest shipping category:
- Records: 6,120
- Unique orders: 5,142
- Sales: $85,490.35
- Gross profit: $56,424.46

## Lead-Time Measure

- Mean: 1,320.84 days
- Median: 1,274 days
- Minimum: 904 days
- Maximum: 1,642 days

Lead time is calculated as Ship Date minus Order Date. It is not
independently verified carrier transit time.
"""

(SUMMARY_DIR / "business_findings.md").write_text(
    textwrap.dedent(business_findings).strip(),
    encoding="utf-8"
)


# ============================================================
# MODEL RESULTS AND METRIC DISCLOSURE
# ============================================================
model_report = """
# Predictive Model Results and Validation

## Selected Algorithm

Tuned Random Forest Regressor.

## Saved Model Metadata

| Metric | Value |
|---|---:|
| Test records | 2,062 |
| MAE | 95.9795 |
| RMSE | 120.7382 |
| R-squared | 0.7929 |

## Model Comparison Output

| Metric | Value |
|---|---:|
| MAE | 139.6707 |
| RMSE | 163.6754 |
| R-squared | 0.6195 |

## Chronological Holdout Output

| Metric | Value |
|---|---:|
| MAE | 179.0003 |
| RMSE | 195.6296 |
| R-squared | -0.1619 |

## Interpretation

The saved model metadata and model comparison output report
different evaluation metrics. The chronological holdout is a
separate evaluation and is reported separately.

The available evidence does not establish the exact reason for
the differences. The evaluation runs must be reconciled before
one metric set is presented as definitive.

MSE was not separately reported in the saved evidence.
"""

(SUMMARY_DIR / "model_results_and_validation.md").write_text(
    textwrap.dedent(model_report).strip(),
    encoding="utf-8"
)


# ============================================================
# MANAGEMENT ACTION PLAN
# ============================================================
action_plan = """
# Management Action Plan

| Area | Recommended action | Evidence required |
|---|---|---|
| Products | Review high-sales and high-profit products | Product-level sales and margin |
| Regions | Investigate regional differences | Regional demand and cost data |
| Shipping | Review shipping-category performance | Carrier pickup and delivery timestamps |
| Lead time | Validate date-derived lead time | Actual logistics event records |
| Factory mapping | Verify product-factory relationships | Manufacturing and dispatch records |
| Capacity | Assess factory feasibility | Capacity and utilization data |
| Cost | Evaluate factory-specific costs | Production and freight costs |
| Model | Reconcile model evaluation metrics | Model run logs and split definitions |
| Dashboard | Use for exploratory management review | Validated and updated data |

## Suggested Implementation

### Immediate
- Review dashboard findings.
- Verify financial and sales summaries.
- Reconcile model evaluation outputs.
- Confirm the meaning of order and ship dates.

### Short Term
- Obtain actual carrier delivery timestamps.
- Validate factory-reference mappings.
- Collect factory capacity and production information.

### Longer Term
- Build a verified factory-level cost dataset.
- Re-evaluate predictive performance using consistent splits.
- Consider operational optimization only after feasibility,
  capacity and cost evidence becomes available.
"""

(SUMMARY_DIR / "management_action_plan.md").write_text(
    textwrap.dedent(action_plan).strip(),
    encoding="utf-8"
)


# ============================================================
# RISK AND ADDITIONAL DATA REQUIREMENTS
# ============================================================
risk_report = """
# Risk, Limitations and Additional Data Requirements

## Main Limitations

- Lead time is derived from Ship Date minus Order Date.
- Lead time is not independently verified carrier transit time.
- Factory mappings are reference mappings, not verified dispatches.
- Haversine distances are straight-line distances, not freight routes.
- Factory-specific savings have not been established.
- Scenario support is not a probability of success.
- Recommendations are indicative, not proven optimal.
- Model evaluation outputs contain unresolved discrepancies.

## Additional Data Required

| Data | Purpose |
|---|---|
| Carrier pickup timestamps | Establish actual shipping start |
| Actual delivery timestamps | Measure transit performance |
| Factory dispatch records | Verify factory-product assignments |
| Factory capacity | Assess production feasibility |
| Factory utilization | Identify constraints |
| Manufacturing costs | Compare production economics |
| Freight rates | Estimate logistics costs |
| Practical freight routes | Estimate route distances |
| Inventory and stockout data | Understand product availability |
| Customer service data | Evaluate service commitments |
| Model evaluation logs | Reconcile model metrics |

No factory savings, delivery improvements or success probabilities
should be claimed without appropriate operational evidence.
"""

(SUMMARY_DIR / "risk_and_data_requirements.md").write_text(
    textwrap.dedent(risk_report).strip(),
    encoding="utf-8"
)


# ============================================================
# POWERPOINT PRESENTATION
# ============================================================
ppt_file = PRESENTATION_DIR / "Nassau_Candy_Executive_Presentation.pptx"

ppt_status = "NOT CREATED"

if importlib.util.find_spec("pptx") is not None:
    from pptx import Presentation
    from pptx.util import Inches, Pt

    prs = Presentation()
    prs.slide_width = Inches(13.333)
    prs.slide_height = Inches(7.5)

    def add_content_slide(title, bullets):
        slide = prs.slides.add_slide(prs.slide_layouts[1])
        slide.shapes.title.text = title
        slide.shapes.title.text_frame.paragraphs[0].font.size = Pt(28)

        frame = slide.placeholders[1].text_frame
        frame.clear()

        for index, bullet in enumerate(bullets):
            paragraph = frame.paragraphs[0] if index == 0 else frame.add_paragraph()
            paragraph.text = bullet
            paragraph.font.size = Pt(19)
            paragraph.space_after = Pt(12)

    cover = prs.slides.add_slide(prs.slide_layouts[0])
    cover.shapes.title.text = "Nassau Candy Distributor"
    cover.placeholders[1].text = (
        "Executive Analytics and Decision-Support Project"
    )

    add_content_slide("Project Overview", [
        "Historical distributor data analysed using analytics methods.",
        "Predictive modelling, clustering and scenario analysis.",
        "Interactive Streamlit dashboard with eight pages.",
        "Focus on evidence-based business investigation."
    ])

    add_content_slide("Business Problem", [
        "Understand product and regional performance.",
        "Review shipping and date-derived lead-time patterns.",
        "Examine profitability and factory-reference information.",
        "Build an accessible management decision-support tool."
    ])

    add_content_slide("Dataset Overview", [
        "Original dataset: 10,194 records and 18 columns.",
        "Processed dataset: 10,194 records and 34 columns.",
        "Data quality, feature engineering and exploratory analysis."
    ])

    add_content_slide("Financial Performance", [
        "Total sales: $141,783.63.",
        "Total cost: $48,340.83.",
        "Gross profit: $93,442.80.",
        "Weighted gross-profit margin: approximately 65.91%."
    ])

    add_content_slide("Product and Regional Findings", [
        "Top product: Wonka Bar - Triple Dazzle Caramel.",
        "Top product sales: $28,485.00.",
        "Pacific region sales: $46,301.53.",
        "Gulf region sales: $22,247.26."
    ])

    add_content_slide("Shipping and Lead Time", [
        "Standard Class: 6,120 records.",
        "Standard Class sales: $85,490.35.",
        "Mean date-derived lead time: 1,320.84 days.",
        "Lead time is not verified carrier transit time."
    ])

    add_content_slide("Predictive Model Results", [
        "Selected algorithm: Tuned Random Forest Regressor.",
        "Metadata: MAE 95.9795, RMSE 120.7382, R-squared 0.7929.",
        "Comparison: MAE 139.6707, RMSE 163.6754, R-squared 0.6195.",
        "Chronological holdout R-squared: -0.1619.",
        "Metric discrepancy remains unresolved."
    ])

    add_content_slide("Recommended Actions", [
        "Review high-sales and high-profit products.",
        "Investigate regional and shipping performance.",
        "Validate actual carrier delivery timestamps.",
        "Verify factory mappings and collect capacity and cost data.",
        "Reconcile model evaluation results."
    ])

    add_content_slide("Risks and Limitations", [
        "Date-derived lead time is not verified carrier transit.",
        "Factory mappings are reference mappings.",
        "Straight-line distances are not freight routes.",
        "Factory savings and scenario success probabilities are unproven.",
        "Recommendations are indicative, not proven optimal."
    ])

    add_content_slide("Conclusion and Next Steps", [
        "Use the dashboard for exploratory management review.",
        "Improve operational data collection.",
        "Validate factory-level capacity and cost evidence.",
        "Reconcile model metrics before operational use."
    ])

    prs.save(ppt_file)
    ppt_status = "CREATED"

else:
    ppt_status = "NOT CREATED: python-pptx is missing"

print("PowerPoint:", ppt_status)


# ============================================================
# README AND SETUP GUIDE
# ============================================================
readme = """
# Nassau Candy Distributor Analytics Project

## Overview

This project analyses historical Nassau Candy Distributor data
using data cleaning, exploratory data analysis, feature engineering,
predictive modelling, clustering, factory-reference analysis,
scenario simulation and decision-support methods.

An interactive Streamlit dashboard is included.

## Project Structure

- data/raw: original input data
- data/processed: processed data and application bundle
- notebooks: analytical Jupyter notebooks
- reports: analytical reports and final delivery documents

## Environment Setup

Create a virtual environment:

    python -m venv .venv

Activate it on Windows:

    .venv\\Scripts\\activate

Install dependencies:

    python -m pip install --upgrade pip
    pip install -r requirements.txt

## Run the Dashboard

    cd data/processed/app_bundle
    streamlit run app.py

Open the local URL shown in the terminal.

## Dashboard Pages

- Home
- Product Performance
- Region Analysis
- Ship Mode Analysis
- Factory Explorer
- Scenario Simulator
- Recommendations
- Risk and Impact

## Data

The original dataset is expected at:

    data/raw/Nassau Candy Distributor.csv

Keep the original raw dataset unchanged.

## Important Limitations

- Lead time is Ship Date minus Order Date.
- It is not independently verified carrier transit time.
- Factory mappings are reference mappings.
- Straight-line distance is not freight route distance.
- Factory-specific savings have not been established.
- Scenario support is not a probability of success.
- Model evaluation outputs contain unresolved discrepancies.

Review the executive summary and model validation report
before using results for management decisions.
"""

readme_file = PROJECT_ROOT / "README.md"
readme_file.write_text(
    textwrap.dedent(readme).strip(),
    encoding="utf-8"
)


# ============================================================
# REQUIREMENTS FILE
# ============================================================
requirements = """
pandas
numpy
matplotlib
seaborn
scikit-learn
scipy
joblib
streamlit
plotly
openpyxl
xlsxwriter
python-pptx
reportlab
python-docx
"""

requirements_file = PROJECT_ROOT / "requirements.txt"
requirements_file.write_text(
    textwrap.dedent(requirements).strip() + "\n",
    encoding="utf-8"
)


# ============================================================
# STREAMLIT APPLICATION USER MANUAL
# ============================================================
manual = """
# Nassau Candy Distributor Dashboard - User Manual

## Starting the Application

1. Open Command Prompt or PowerShell.
2. Navigate to the application folder:

       cd data/processed/app_bundle

3. Start the dashboard:

       streamlit run app.py

4. Open the local browser address displayed by Streamlit.

## Dashboard Pages

### Home
Displays the project overview and summary information.

### Product Performance
Explore product-level sales and performance patterns.

### Region Analysis
Review regional sales and profitability.

### Ship Mode Analysis
Explore shipping-category records and related measures.

### Factory Explorer
Review product-to-factory reference information.
Mappings are not confirmed historical dispatch records.

### Scenario Simulator
Explore supported scenario information.
Scenario support does not establish actual savings,
operational feasibility or probability of success.

### Recommendations
Review indicative analytical priorities.
Recommendations are not proven optimal decisions.

### Risk and Impact
Review risks and analytical limitations.

## Filters and Downloads

Use available filters to narrow the displayed information.
Use CSV or Excel download controls where provided.

## Interpretation Guidance

- Check filter selections before interpreting charts.
- Distinguish sales, cost and gross profit.
- Lead time is derived from order and ship dates.
- Factory-specific savings are not established.
- Model metrics have an unresolved discrepancy.

## Troubleshooting

1. Confirm that the virtual environment is activated.
2. Install dependencies from requirements.txt.
3. Confirm that app.py exists.
4. Confirm that required data and model files are present.
5. Review the terminal error message.

Do not delete data or model files to resolve an error
without first creating a backup.
"""

manual_file = DOCUMENTATION_DIR / "application_user_manual.md"
manual_file.write_text(
    textwrap.dedent(manual).strip(),
    encoding="utf-8"
)


# ============================================================
# FRESH ENVIRONMENT TEST SCRIPT
# ============================================================
# This creates a test script. It does not claim that a clean
# environment has already been tested.

fresh_test = r'''
from pathlib import Path
import importlib.util
import ast

PROJECT_ROOT = Path(
    r"C:\Users\hp\Nassau_Candy_Final_Submission"
)
APP_ROOT = PROJECT_ROOT / "data" / "processed" / "app_bundle"
APP_FILE = APP_ROOT / "app.py"

packages = [
    "pandas", "numpy", "sklearn", "joblib",
    "streamlit", "plotly", "openpyxl"
]

failed = []

print("=" * 60)
print("FRESH ENVIRONMENT VALIDATION")
print("=" * 60)

for package in packages:
    available = importlib.util.find_spec(package) is not None
    print(package, "PASS" if available else "FAIL")
    if not available:
        failed.append(package)

if APP_FILE.exists():
    try:
        ast.parse(APP_FILE.read_text(encoding="utf-8"))
        print("Application syntax: PASS")
    except Exception as error:
        print("Application syntax: FAIL", error)
        failed.append("application syntax")
else:
    print("Application file: FAIL")
    failed.append("app.py")

for module in [
    "prediction_utils.py",
    "feature_utils.py",
    "scenario_utils.py",
    "recommendation_utils.py"
]:
    candidates = [
        APP_ROOT / "src" / module,
        APP_ROOT / module
    ]
    found = any(path.exists() for path in candidates)
    print(module, "PASS" if found else "REVIEW REQUIRED")

print("STATUS:", "PASS" if not failed else "FAIL")
if failed:
    print("Failed checks:", failed)
'''

fresh_test_file = VALIDATION_DIR / "fresh_environment_test.py"
fresh_test_file.write_text(
    textwrap.dedent(fresh_test).strip(),
    encoding="utf-8"
)


# ============================================================
# CURRENT PROJECT VALIDATION
# ============================================================
# Checks file presence and Python syntax in the current
# environment. A separate clean environment test is still needed.

checks = []

def add_check(name, passed, detail=""):
    checks.append({
        "check": name,
        "status": "PASS" if passed else "REVIEW REQUIRED",
        "detail": str(detail)
    })

add_check("Project root", PROJECT_ROOT.exists(), PROJECT_ROOT)
add_check("Application entry point", APP_FILE.exists(), APP_FILE)
add_check(
    "Raw dataset",
    (RAW_DIR / "Nassau Candy Distributor.csv").exists()
)
add_check("Executive summary", summary_file.exists())
add_check("README", readme_file.exists())
add_check("Requirements", requirements_file.exists())
add_check("PowerPoint", ppt_file.exists(), ppt_status)
add_check("User manual", manual_file.exists())
add_check("Fresh test script", fresh_test_file.exists())

if APP_FILE.exists():
    try:
        ast.parse(APP_FILE.read_text(encoding="utf-8"))
        add_check("Application syntax", True)
    except Exception as error:
        add_check("Application syntax", False, error)

for module in [
    "prediction_utils.py",
    "feature_utils.py",
    "scenario_utils.py",
    "recommendation_utils.py"
]:
    candidates = [
        APP_ROOT / "src" / module,
        APP_ROOT / module
    ]
    found = any(path.exists() for path in candidates)
    add_check(f"Application module: {module}", found)

passed = sum(item["status"] == "PASS" for item in checks)
review = len(checks) - passed

validation_file = VALIDATION_DIR / "project_validation_results.json"
validation_file.write_text(
    json.dumps(checks, indent=4),
    encoding="utf-8"
)

print("\nPROJECT VALIDATION")
for item in checks:
    print(
        item["status"], "|",
        item["check"], "|",
        item["detail"]
    )

print("\nPassed:", passed)
print("Review required:", review)


# ============================================================
# DELIVERY INVENTORY
# ============================================================
inventory = []

for path in DELIVERY_DIR.rglob("*"):
    if path.is_file():
        inventory.append({
            "file": str(path.relative_to(PROJECT_ROOT)),
            "size_kb": round(path.stat().st_size / 1024, 2)
        })

inventory_file = DELIVERY_DIR / "delivery_inventory.json"
inventory_file.write_text(
    json.dumps(inventory, indent=4),
    encoding="utf-8"
)


# ============================================================
# FINAL ZIP HANDOVER PACKAGE
# ============================================================
# Excludes temporary environments and cache folders.
# The ZIP file itself is excluded from its own archive.

zip_file = PACKAGE_DIR / "Nassau_Candy_Final_Handover.zip"

if zip_file.exists():
    zip_file.unlink()

excluded = {
    ".venv", "__pycache__", ".ipynb_checkpoints", ".git"
}

zip_errors = []

with zipfile.ZipFile(
    zip_file, "w", compression=zipfile.ZIP_DEFLATED
) as archive:

    for path in PROJECT_ROOT.rglob("*"):
        if not path.is_file():
            continue

        relative = path.relative_to(PROJECT_ROOT)

        if any(part in excluded for part in relative.parts):
            continue

        if path.resolve() == zip_file.resolve():
            continue

        try:
            archive.write(
                path,
                arcname=str(Path(PROJECT_ROOT.name) / relative)
            )
        except Exception as error:
            zip_errors.append((str(relative), str(error)))

print("\nZIP errors:", len(zip_errors))
for item in zip_errors:
    print("REVIEW REQUIRED:", item)


# ============================================================
# FINAL DELIVERY SUMMARY
# ============================================================
print("\n" + "=" * 70)
print("FINAL DELIVERY SUMMARY")
print("=" * 70)

print("Executive summary:", summary_file.exists())
print("Business findings:", (SUMMARY_DIR / "business_findings.md").exists())
print("Model report:", (SUMMARY_DIR / "model_results_and_validation.md").exists())
print("Action plan:", (SUMMARY_DIR / "management_action_plan.md").exists())
print("Risk report:", (SUMMARY_DIR / "risk_and_data_requirements.md").exists())
print("PowerPoint:", ppt_status)
print("README:", readme_file.exists())
print("Requirements:", requirements_file.exists())
print("User manual:", manual_file.exists())
print("Fresh test script:", fresh_test_file.exists())
print("Validation report:", validation_file.exists())
print("Handover ZIP:", zip_file.exists())

print("\nCurrent-environment checks passed:", passed)
print("Current-environment checks requiring review:", review)

print("\nOutstanding items:")
print("- Run the fresh-environment test in a clean virtual environment.")
print("- Reconcile model evaluation metrics.")
print("- Verify all references and source files before submission.")
print("- Review the ZIP contents before sharing externally.")

print("\nDelivery folder:", DELIVERY_DIR)
print("Handover package:", zip_file)

if review == 0 and ppt_file.exists() and not zip_errors:
    print("\nSTATUS: DELIVERY FILES CREATED - FRESH TEST STILL REQUIRED")
else:
    print("\nSTATUS: DELIVERY CREATED - REVIEW ITEMS REMAIN")

print("=" * 70)
```
