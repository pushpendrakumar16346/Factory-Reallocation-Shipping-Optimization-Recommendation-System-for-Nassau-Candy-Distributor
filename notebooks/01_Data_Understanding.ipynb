```python
# ==========================================
# CREATE PROJECT NOTEBOOK FOLDER
# ==========================================

from pathlib import Path

current_path = Path.cwd()

print("Current Location:", current_path)

# Locate project folder
if current_path.name == "Nassau_Candy_Final_Submission":
    project_root = current_path
elif (current_path / "Nassau_Candy_Final_Submission").exists():
    project_root = current_path / "Nassau_Candy_Final_Submission"
elif current_path.name == "notebooks":
    project_root = current_path.parent
else:
    project_root = current_path

# Create notebooks folder
notebooks_folder = project_root / "notebooks"
notebooks_folder.mkdir(parents=True, exist_ok=True)

print("\nNotebooks Folder:", notebooks_folder)
print("Folder Created:", notebooks_folder.exists())
```
